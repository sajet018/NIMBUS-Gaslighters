Importing required libraries.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

Displaying image using cv2.imshow().

```python
raw = cv2.imread("cvtest.jpeg")

if raw is None:
    print("Image not found. Check filename/path.")
else:
    cv2.imshow("raw", raw)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

Displaying image using plt.imshow().

```python
img_rgb = cv2.cvtColor(raw, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()

print(raw.shape)
print(raw[10, 10])
```

Displaying a red square and a green line.

```python
sample = np.zeros((100, 100, 3), dtype=np.uint8)
print(sample.shape)

sample[10:40, 10:40] = [255, 0, 0]

sample[49:52, :] = [0, 255, 0]

plt.imshow(sample)
plt.axis("off")
plt.show()
```

Creating a boolean mask and displaying it.

```python
mask = np.zeros((100, 100), dtype = bool)
mask[20:30, 20:30] = True
plt.imshow(mask, cmap = "gray")
plt.axis("off")
```

Applying the boolean mask to sample.

```python
sample[mask] = (255, 255, 255)
plt.imshow(sample)
```

Creating another image.

```python
img = np.zeros((100, 100, 3), dtype=np.uint8)
print(img.shape)

img[0:100, 90:100] = [255, 0, 0]

img[0:10, 0:100] = [0, 255, 0]

img[30:70, 30:70] = [0, 0, 255]

plt.imshow(img)
plt.axis("off")
plt.show()
```

Using boolean masks to count pixels within a certain range in a random integer array.

```python
arr = np.random.randint(0, 256, size = (50, 50), dtype = np.uint8)

mask_200 = arr>200

count_200 = np.sum(mask_200)
print("Pixels > 200:", count_200)

mask_255 = arr==200

count_255 = np.sum(mask_255)
print("Pixels = 255:", count_255)
```

Creating blue, green and red channels and displaying them with the original image.

```python
blue = raw[:, :, 0]
green = raw[:, :, 1]
red = raw[:, :, 2]

plt.subplots(nrows = 2, ncols = 2)

plt.subplot(221)
plt.imshow(img_rgb)
plt.axis("off")
plt.title("Image")

plt.subplot(222)
plt.imshow(blue, cmap="grey")
plt.axis("off")
plt.title("Blue Channel")

plt.subplot(223)
plt.imshow(green, cmap="grey")
plt.axis("off")
plt.title("Green Channel")

plt.subplot(224)
plt.imshow(red,  cmap="grey")
plt.axis("off")
plt.title("Red Channel")
plt.show()

print("Blue Channel Mean = ", np.mean(blue))
print("Green Channel Mean = ", np.mean(green))
print("Red Channel Mean = ", np.mean(red))
```
